```python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "Qwen/Qwen2.5-0.5B-Instruct"

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained(model_name)

print("Model dtype:", model.dtype)
print("Model device:", model.device)
print("Model generation config:", model.generation_config)
```

```python


prompt = "What is the return policy at nordvik electronics?"
messages = [
    {"role": "system", "content": "You are Qwen, created by Alibaba Cloud. You are a helpful assistant."},
    {"role": "user", "content": prompt}
]
text_false = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=False
)

text_true = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

text_without_system = tokenizer.apply_chat_template(
    conversation=[{"role": "user", "content": prompt}],
    tokenize=False,
    add_generation_prompt=True,
)

text_with_tokenization = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True
)

print("Prompt with generation prompt: \n", text_true)
print("########\n")
print("Prompt without generation prompt: \n", text_false)
print("########\n")
print("Prompt without system prompt: \n", text_without_system)
print("########\n")
print("Prompt with tokenization: \n", text_with_tokenization)
```

```python
model_inputs = tokenizer([text_true], return_tensors="pt").to(model.device)

generated_ids = model.generate(
    **model_inputs,
    max_new_tokens=100,
    do_sample=False
)
print("Generated ids: \n", generated_ids)
```

```python
generated_ids = [
    output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
]

response = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
```
